```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = np.load('../data/bulkP.npz')
temps = df["temps"]
F0 = df["F0"]
F_swave = df["F_swave"]
F_dwave = df["F_dwave"]
F_px = df["F_px"]
F_py = df["F_py"]

plt.figure(1)
plt.clf()

# plt.plot(temps, np.real(F_swave), linewidth=1, label=r"$s$-wave")
# plt.plot(temps, np.real(F_dwave), linewidth=1, label=r"$d$-wave")
plt.plot(temps, np.real(F_px), linewidth=1, label=r"$p_x$-wave", color='black')
plt.plot(temps, np.imag(F_py), "--", linewidth=1, label=r"$p_y$-wave", color='red')
plt.plot(np.arange(0.4,0.61,0.01), np.zeros(21), color='black')
plt.plot(np.arange(0.4,0.61,0.01), np.zeros(21),'--', color='red')

plt.ylim(0,0.25)
plt.xlim(0,0.6)
plt.xlabel("T", fontsize=15)
plt.legend(loc="best", fontsize=12)
plt.tight_layout()
plt.show()
print(temps[np.where(df["converged"]==False)])
```
