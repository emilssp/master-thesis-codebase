```python
import matplotlib.pyplot as plt
import numpy as np

df = np.load('../data/SFP.npz')
temps = df["temp"]
F0 = df["F0"]
F_swave = df["F_swave"]
F_dwave = df["F_dwave"]
F_px = df["F_px"]
F_py = df["F_py"]

plt.figure(1)
plt.clf()
plt.plot(temps, np.real(F_px), linewidth=1, label=r"$p_x$-wave")
plt.plot(temps, np.imag(F_py), "--", linewidth=1, label=r"$p_y$-wave")
# plt.ylim(0,0.2)
plt.xlim(0,0.6)
plt.xlabel("T", fontsize=15)
plt.legend(loc="best", fontsize=12)
plt.tight_layout()
plt.show()
```
