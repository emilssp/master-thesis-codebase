```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = np.load('../data/bulkP.npz')
temps = df["temps"]
F0 = df["F0"]
F_swave = df["F_swave"]
F_dwave = df["F_dwave"]
F_px = df["F_px"]
F_py = df["F_py"]

plt.figure(1)
plt.clf()

# plt.plot(temps, np.real(F_swave), linewidth=1, label=r"$s$-wave")
# plt.plot(temps, np.real(F_dwave), linewidth=1, label=r"$d$-wave")
plt.plot(temps, np.real(F_px), linewidth=1, label=r"$p_x$-wave")
plt.plot(temps, np.imag(F_py), "--", linewidth=1, label=r"$p_y$-wave")
# plt.ylim(0,0.25)
# plt.xlim(0,0.6)
plt.xlabel("T", fontsize=15)
plt.legend(loc="best", fontsize=12)
plt.tight_layout()
plt.show()
```

```python
from turtle import color

import numpy as np
import matplotlib.pyplot as plt
from pyparsing import col

df = np.load('../data/bulkTSC.npz')
temps = df["temps"]
F0 = df["F0"]
F_swave = df["F_swave"]
F_dwave = df["F_dwave"]
F_px = df["F_px"]
F_py = df["F_py"]
Fuu_px = df["Fuu_px"]
Fuu_py = df["Fuu_py"]
Fdd_px = df["Fdd_px"]
Fdd_py = df["Fdd_py"]

plt.figure(1)
plt.clf()

# plt.plot(temps, np.real(F_swave), linewidth=1, label=r"$s$-wave")
# plt.plot(temps, np.real(F_dwave), linewidth=1, label=r"$d$-wave")
plt.plot(temps, np.real(Fuu_px), linewidth=1, label=r"$uu:p_x$-wave", color='b')
plt.plot(temps, np.imag(Fuu_py), linewidth=1, label=r"$uu:p_y$-wave", color='g')
plt.plot(temps, np.real(Fdd_px), "--", linewidth=1, label=r"$dd:p_x$-wave", color='r')
plt.plot(temps, np.imag(Fdd_py), "--", linewidth=1, label=r"$dd:p_y$-wave", color='k')
# plt.ylim(0,0.25)
# plt.xlim(0,0.6)
plt.xlabel("T", fontsize=15)
plt.legend(loc="best", fontsize=12)
plt.tight_layout()
plt.show()
```
