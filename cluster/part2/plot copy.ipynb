```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = np.load('./data/SP.npz')
temps = df["temps"]
F0 = df["F0"]
F_swave = df["F_swave"]
F_dwave = df["F_dwave"]
F_px = df["F_px"]
F_py = df["F_py"]

plt.figure(1)
plt.clf()
# plt.plot(temps, np.real(F0))
plt.plot(temps, np.abs(F_px), linewidth=1, label=r"$p_x$-wave", color='black')
plt.plot(temps, np.abs(F_py), "--", linewidth=1, label=r"$p_y$-wave", color='red')
plt.ylim(0,0.05)
plt.xlim(0,0.2)
plt.xlabel("T", fontsize=15)
plt.legend(loc="best", fontsize=12)
plt.tight_layout()
plt.show()
print(temps[np.where(df["converged"]==False)])
```

```python
df = np.load('./data/SFP.npz')
temps = df["temps"]
F0 = df["F0"]
F_swave = df["F_swave"]
F_dwave = df["F_dwave"]
F_px = df["F_px"]
F_py = df["F_py"]

plt.figure(1)
plt.clf()
plt.title("SFP - opposite spin")
plt.plot(temps, -np.real(F_px), linewidth=1, label=r"$p_x$-wave", color='black')
plt.plot(temps, np.imag(F_py), "--", linewidth=1, label=r"$p_y$-wave", color='red')
plt.ylim(0,0.05)
plt.xlim(0,0.2)
plt.xlabel("T", fontsize=15)
plt.legend(loc="best", fontsize=12)
plt.tight_layout()
plt.show()

print(temps[np.where(np.isclose(temps,0.2))])
print(F_px[np.where(np.isclose(temps,0.2))])
```
