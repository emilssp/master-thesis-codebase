```python
import numpy as np
import pandas as pd

from constants import *
from self_consistency_k import bdg_sc_full_k
```

```python
t=1
V_prime= 1.5

h = np.array([0, 0, 0])
Nx, Ny = 75, 75

atol = 1e-5
rtol = 1e-3
maxiter=200
mu = -2

temps = np.linspace(0.0001, 0.1, 1)

rows = []

for idx, temp in enumerate(temps):
    print(f"Running job {idx+1} out of {len(temps)}")

    V_prime = 1.5 * t 

    out = bdg_sc_full_k(
        t, mu, Nx, Ny, temp, 
        V_prime=V_prime,
        atol=atol, rtol=rtol, maxiter=maxiter,
        Fuu_init=np.array([0.1, -0.1, 0.1j, -0.1j]),
        Fdd_init=np.array([0.1, -0.1, 0.1j, -0.1j])
    )

    rows.append({
        "idx": idx,
        "temp": temp,
        "Fuu_px": out.Fuu_px,
        "Fuu_py": out.Fuu_py,
        "Fdd_px": out.Fdd_px,
        "Fdd_py": out.Fdd_py,
    })

df = pd.DataFrame(rows)
```
